## 1. Linear Algebra

### 1.1 Vectors

```python

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

dimension = rng.integers(3, 6)
vector_a = rng.integers(-10, 10, size=dimension).astype(float)
vector_b = rng.integers(-10, 10, size=dimension).astype(float)
scalar = rng.uniform(0.5, 3.0)

vector_sum = vector_a + vector_b
vector_diff = vector_a - vector_b
scaled_vector = scalar * vector_a
norm_a = np.linalg.norm(vector_a)
norm_b = np.linalg.norm(vector_b)

print("Vector A:", vector_a)
print("Vector B:", vector_b)
print("Dimension:", dimension)
print("A + B:", vector_sum)
print("A - B:", vector_diff)
print("Scalar:", scalar)
print("scalar * A:", scaled_vector)
print("Norm of A:", norm_a)
print("Norm of B:", norm_b)
```

```python

plot_vector_a = vector_a[:2]
plot_vector_b = vector_b[:2]

figure_vectors, axis_vectors = plt.subplots(figsize=(6, 6))
axis_vectors.quiver(0, 0, plot_vector_a[0], plot_vector_a[1], angles='xy', scale_units='xy', scale=1, color='royalblue', label='Vector A')
axis_vectors.quiver(0, 0, plot_vector_b[0], plot_vector_b[1], angles='xy', scale_units='xy', scale=1, color='crimson', label='Vector B')

axis_limit = np.abs(np.concatenate([plot_vector_a, plot_vector_b])).max() + 2
axis_vectors.set_xlim(-axis_limit, axis_limit)
axis_vectors.set_ylim(-axis_limit, axis_limit)
axis_vectors.axhline(0, color='gray', linewidth=0.5)
axis_vectors.axvline(0, color='gray', linewidth=0.5)
axis_vectors.set_title('Vector Visualization (first two dimensions)')
axis_vectors.legend()
axis_vectors.grid(True)
plt.show()
```

### 1.4 Eigenvalues and Eigenvectors (Intuition)

```python

eigen_size = rng.integers(2, 4)
matrix_e = rng.integers(-6, 6, size=(eigen_size, eigen_size)).astype(float)

eigenvalues, eigenvectors = np.linalg.eig(matrix_e)

print("Matrix E:")
print(matrix_e)
print("Eigenvalues:")
print(eigenvalues)
print("Eigenvectors (columns):")
print(eigenvectors)

for index in range(eigen_size):
    left_side = matrix_e @ eigenvectors[:, index]
    right_side = eigenvalues[index] * eigenvectors[:, index]
    print("Eigenvector index:", index)
    print("E @ v:", left_side)
    print("lambda * v:", right_side)
    print("Difference close to zero:", np.allclose(left_side, right_side))
```

## 2. Calculus

### 2.1 Derivatives

```python

import sympy as sp

x = sp.symbols('x')

degree = rng.integers(2, 5)
coefficients = rng.integers(-5, 5, size=degree + 1)
while coefficients[0] == 0:
    coefficients[0] = rng.integers(-5, 5)

polynomial_expr = sum(int(coefficients[i]) * x**(degree - i) for i in range(degree + 1))
derivative_expr = sp.diff(polynomial_expr, x)

evaluation_point = rng.uniform(-3, 3)
derivative_value = derivative_expr.subs(x, evaluation_point)

print("Polynomial function f(x):", polynomial_expr)
print("Derivative f'(x):", derivative_expr)
print("Evaluation point:", evaluation_point)
print("f'(evaluation point):", derivative_value)
```

```python

polynomial_func = sp.lambdify(x, polynomial_expr, 'numpy')
derivative_func = sp.lambdify(x, derivative_expr, 'numpy')

x_range = np.linspace(evaluation_point - 3, evaluation_point + 3, 200)
y_range = polynomial_func(x_range)
tangent_y_at_point = polynomial_func(evaluation_point)
tangent_line = tangent_y_at_point + derivative_value * (x_range - evaluation_point)

figure_derivative, axis_derivative = plt.subplots(figsize=(7, 5))
axis_derivative.plot(x_range, y_range, color='royalblue', label='f(x)')
axis_derivative.plot(x_range, tangent_line, color='crimson', linestyle='--', label='Tangent Line')
axis_derivative.scatter([evaluation_point], [tangent_y_at_point], color='black', zorder=5)
axis_derivative.set_title('Function and Tangent Line at x = ' + str(round(evaluation_point, 2)))
axis_derivative.set_xlabel('x')
axis_derivative.set_ylabel('f(x)')
axis_derivative.legend()
axis_derivative.grid(True)
plt.show()
```

### 2.2 Gradients

```python

x1, x2 = sp.symbols('x1 x2')

coeff_a = rng.integers(1, 6)
coeff_b = rng.integers(1, 6)
coeff_c = rng.integers(1, 6)

multivariable_expr = int(coeff_a) * x1**2 + int(coeff_b) * x2**2 + int(coeff_c) * x1 * x2

gradient_expr = [sp.diff(multivariable_expr, var) for var in (x1, x2)]

point_x1 = rng.uniform(-3, 3)
point_x2 = rng.uniform(-3, 3)

gradient_value = [component.subs({x1: point_x1, x2: point_x2}) for component in gradient_expr]

print("Function f(x1, x2):", multivariable_expr)
print("Gradient vector:", gradient_expr)
print("Evaluation point:", (point_x1, point_x2))
print("Gradient at evaluation point:", gradient_value)
```

```python

multivariable_func = sp.lambdify((x1, x2), multivariable_expr, 'numpy')

grid_range = np.linspace(-4, 4, 100)
grid_x1, grid_x2 = np.meshgrid(grid_range, grid_range)
grid_z = multivariable_func(grid_x1, grid_x2)

figure_gradient, axis_gradient = plt.subplots(figsize=(7, 6))
contour_plot = axis_gradient.contourf(grid_x1, grid_x2, grid_z, levels=20, cmap='viridis')
figure_gradient.colorbar(contour_plot, ax=axis_gradient)

axis_gradient.quiver(point_x1, point_x2, float(gradient_value[0]), float(gradient_value[1]), color='white', scale=50, width=0.006)
axis_gradient.scatter([point_x1], [point_x2], color='red', zorder=5)
axis_gradient.set_title('Function Contours and Gradient Direction')
axis_gradient.set_xlabel('x1')
axis_gradient.set_ylabel('x2')
plt.show()
```

### 2.3 Chain Rule (for Backpropagation Intuition)

```python

weight, bias, input_value, target_value, z_symbol, a_symbol = sp.symbols('weight bias input_value target_value z_symbol a_symbol')

linear_output = weight * input_value + bias
activation_function = 1 / (1 + sp.exp(-z_symbol))
loss_function = (a_symbol - target_value)**2

derivative_loss_wrt_activated = sp.diff(loss_function, a_symbol)
derivative_activated_wrt_linear = sp.diff(activation_function, z_symbol)
derivative_linear_wrt_weight = sp.diff(linear_output, weight)

activated_output = activation_function.subs(z_symbol, linear_output)
loss_expr = loss_function.subs(a_symbol, activated_output)

chain_rule_gradient = (
    derivative_loss_wrt_activated.subs(a_symbol, activated_output)
    * derivative_activated_wrt_linear.subs(z_symbol, linear_output)
    * derivative_linear_wrt_weight
)
direct_gradient = sp.diff(loss_expr, weight)

numeric_values = {
    weight: rng.uniform(-2, 2),
    bias: rng.uniform(-2, 2),
    input_value: rng.uniform(-2, 2),
    target_value: rng.uniform(0, 1)
}

chain_rule_numeric = chain_rule_gradient.subs(numeric_values)
direct_numeric = direct_gradient.subs(numeric_values)

print("Linear output (z = w*x + b):", linear_output)
print("Activated output (sigmoid):", activated_output)
print("Loss function:", loss_expr)
print("dLoss/dActivated:", derivative_loss_wrt_activated)
print("dActivated/dLinear:", derivative_activated_wrt_linear)
print("dLinear/dWeight:", derivative_linear_wrt_weight)
print("Chain rule gradient dLoss/dWeight:", sp.simplify(chain_rule_gradient))
print("Direct gradient dLoss/dWeight:", sp.simplify(direct_gradient))
print("Numeric values used:", numeric_values)
print("Chain rule numeric result:", chain_rule_numeric)
print("Direct differentiation numeric result:", direct_numeric)
print("Results match:", sp.simplify(chain_rule_gradient - direct_gradient) == 0)
```

```python

loss_func_of_weight = sp.lambdify(weight, loss_expr.subs({bias: numeric_values[bias], input_value: numeric_values[input_value], target_value: numeric_values[target_value]}), 'numpy')
gradient_func_of_weight = sp.lambdify(weight, direct_gradient.subs({bias: numeric_values[bias], input_value: numeric_values[input_value], target_value: numeric_values[target_value]}), 'numpy')

weight_range = np.linspace(float(numeric_values[weight]) - 3, float(numeric_values[weight]) + 3, 200)
loss_values = loss_func_of_weight(weight_range)
current_loss = loss_func_of_weight(float(numeric_values[weight]))
current_gradient = gradient_func_of_weight(float(numeric_values[weight]))
tangent_line_loss = current_loss + current_gradient * (weight_range - float(numeric_values[weight]))

figure_chain, axis_chain = plt.subplots(figsize=(7, 5))
axis_chain.plot(weight_range, loss_values, color='royalblue', label='Loss(weight)')
axis_chain.plot(weight_range, tangent_line_loss, color='crimson', linestyle='--', label='Tangent via Chain Rule')
axis_chain.scatter([float(numeric_values[weight])], [current_loss], color='black', zorder=5)
axis_chain.set_title('Loss Curve and Chain Rule Gradient at Current Weight')
axis_chain.set_xlabel('weight')
axis_chain.set_ylabel('loss')
axis_chain.legend()
axis_chain.grid(True)
plt.show()
```
